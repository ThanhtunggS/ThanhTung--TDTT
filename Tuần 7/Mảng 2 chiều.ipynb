trả về cột (chỉ số bắt đầu từ 0) có tổng cột lớn nhất trong ma trận n x m
Viết hàm column_with_max_sum(mat) trả về chỉ số bắt đầu từ 0 của cột có tổng lớn nhất.
Nếu có nhiều cột cùng tổng lớn nhất, trả về cột có chỉ số nhỏ nhất - Dùng vòng lặp để tính từng cột - ko dùng tuple hoặc dictionary

```python
# C1: Tính trực tiếp

#Hàm nhập 
n, m = map(int, input().split())
matran = []
for _ in range(n):
    while True:
        hang = list(map(int, input().split()))
        if len(hang) == m:
            matran.append(hang)
            break
print (matran)

#Hàm tính max
def column_with_max_sum (mat):
    lst_tong = []
    for x in range(len(mat[1])):
        lst = []
        for i in mat:
            lst.append(i[x])
        lst_tong.append(sum(lst))
    return (max(lst_tong))
print (column_with_max_sum(matran))
```

```python
#C2: Tạo ra các list collumn riêng sau đó tìm list có giá trị lớn nhất

#Hàm nhập 
n, m = map(int, input().split())
matran = []
for _ in range(n):
    while True:
        hang = list(map(int, input().split()))
        if len(hang) == m:
            matran.append(hang)
            break
print (matran)

#Hàm tính max
def column_with_max_sum (mat):
    col = []
    for i in range (len(mat[0])):
        col.append([])
      
    for i in range(len(mat[0])):
        for x in mat:
            col[i].append(x[i])
    k = sum(col[1])
    for j in col:
        if sum(j) > k:
            k = sum(j)
    for l in col:
        if sum(l) == k:
            return (l)           
print (column_with_max_sum(matran))                
            
```
