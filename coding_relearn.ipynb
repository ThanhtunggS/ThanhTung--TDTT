```python
def binary_search (arr, x):
    right = len(arr) - 1
    left = 0
    while left <= right: #Phải là left bé hơn HOẶC BẰNG right
        mid = (right + left)//2
        if x == arr[mid]:
            return mid
        if x > arr[mid]:
            left = mid + 1
        if x < arr[mid]:
            right = mid - 1
    return -1
x = int(input())
arr = list(map(int,input().split()))
print (binary_search(arr, x))   
```

Sử dụng thuật toán tìm kiếm nhị phân để tìm x.

Kết quả: In ra chỉ số (index) của x nếu tìm thấy, ngược lại in ra -1.

```python
#C1: Dùng quick sort:

def quick_sort (arr):
    if len(arr) <= 1:
        return arr
    else:
        mid = arr[len(arr)//2]
        left = [i for i in arr if i < mid]
        right = [i for i in arr if i > mid]
        return quick_sort(left) + [mid] + quick_sort(right)

def binary_search (arr, x):
    left = 0
    right = len(arr) - 1
    while left <= right:
        mid = (left + right)//2
        if arr[mid] == x:
            return mid
        if x > mid:
            left = mid + 1
        if x < mid:
            right = mid - 1
    return -1

arr = list(map(int,input().split()))
x = int(input())

arr = quick_sort(arr)
print (arr)
print (binary_search(arr, x))                  

```

#### Quy trình đệ quy trong máy

Hàm quick_sort (arr): -> Tạo ra left, right, middle

return (quick_sort(left))

Vấn đề: quick_sort(left) còn chưa đc tính

--> Máy bắt đầu tính quick_sort(left)

left có [2, 1, 3] --> Quick sort ra đc: left: [1], middle: [2], right: [3]

[1], [2] và [3] chỉ có 1 phần tử -> được return làm giá trị gốc tại hàm return arr

VD: quick_sort(left) = quick_sort([1]) + quick_sort([2]) + quick_sort([3])

quick_sort([1]) = [1] (return arr)
quick_sort([2]) = [2]
quick_sort([3]) = [3]

-> quick_sort(left) = [1] + [2] + [3] = [1, 2, 3]

Sau khi đã xác nhận được [1], [2] và [3] là giá trị gốc, quick_sort(left) = [1] + [2] + [3] = [1, 2, 3]

### 2: Thuật toán Đệ quy

Viết hàm ĐỆ QUY reverse_string (s) để đảo ngược một chuỗi ký tự đầu vào

```python
def reverse_string (s):
    if len(s) <= 1:
        return s
    else:
        return (reverse_string(s[1:])) + s[0]
s = input()
print (reverse_string(s))        
```

#### Hướng dẫn giải đệ quy:

tích tiểu thành đại, biến đổi một phần tử nhỏ trước.

Sau đó, lặp lại phép biến đổi đấy với chính kết quả của bài.

VD: Ở bài trên  reverse_string(s[1:]) + s[0] 

Cắt s ra thành s[1:] + s[0], ta lấy đi s[0] của s và giữ lại s[1:] để cắt tiếp
 
sau đó s[1:] sẽ liên tục bị cắt (và lấy đi phần tử đầu tiên) cho đến khi gặp điều kiện len(s) <= 1. Uwa, đây là điểm dừng.


## Tư duy khi làm đệ quy:

Mấu chốt nằm ở hàm return MÀ GỌI LẠI HÀM ĐANG DÙNG (Tạm gọi đây là return đệ quy).

MẤU CHỐT NHƯ SAU:

Return cuối sẽ dẫn đến một hàm con, mà hàm con đó sẽ được xử lý giống như cách mà hàm cha được xử lý qua return

VD: return (reverse_string (s[1:])) + s[0]

Ở hàm cha, return này có yếu tố đệ quy là (reverse_string (s[1:])), nó cắt hàm cha, chỉ lấy phần tử sau 1

Thì hàm con cũng bị như vậy: nó sẽ bị cắt đi, lấy phần tử sau 1

Lặp lại cho đến khi hàm cháu chắt chít chỉ còn đúng 1 phần tử, thì sẽ khởi động hàm return s
