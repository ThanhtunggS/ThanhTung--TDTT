* **Output:** Tổng $1 + 2 + 3 + ... + N$.

### Bài 2: Tính giai thừa của N (Cơ bản)
**Đề bài:** Viết hàm đệ quy `factorial(n)` để tính giai thừa của số nguyên không âm $N$ ($N! = 1 \times 2 \times ... \times N$). Quy ước $0! = 1$.
* **Input:** Một số nguyên không âm $N$.
* **Output:** Giá trị $N!$.
* **Ví dụ:** `factorial(4)` trả về `24`.

```python
def factorial(n):
    if n == 1:
        return n
    else:
        return n * factorial(n-1)
```

### Bài 3: Tìm số Fibonacci thứ N (Cơ bản - Vừa phải)
**Đề bài:** Dãy Fibonacci được định nghĩa là: $F(0) = 0$, $F(1) = 1$, và $F(n) = F(n-1) + F(n-2)$ với $n \ge 2$. Viết hàm đệ quy `fibonacci(n)` để tìm số Fibonacci thứ $n$.
* **Input:** Một số nguyên không âm $n$.
* **Output:** Số Fibonacci thứ $n$.
* **Ví dụ:** `fibonacci(6)` trả về `8`.

```python
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    else:
        return fibonacci(n- 1) + fibonacci(n-2)    
n = int(input())
print (fibonacci(n))
```

### Bài 1: Sắp xếp nổi bọt (Bubble Sort) - Dễ
**Đề bài:** Viết hàm `bubble_sort(arr)` sắp xếp danh sách `arr` theo thứ tự tăng dần bằng thuật toán Bubble Sort. Trả về mảng đã sắp xếp.
* **Input:** Một danh sách số nguyên `arr`.
* **Output:** Danh sách đã được sắp xếp tăng dần.
* **Ví dụ:** `bubble_sort([5, 1, 4, 2, 8])` -> `[1, 2, 4, 5, 8]`.

Bản chất: Đẩy phần tử lớn nhất về cuối mảng bằng cách liên tục so sánh hai phần tử kề nhau. Nếu phần tử đứng trước lớn hơn phần tử đứng sau, chúng sẽ hoán đổi vị trí.

```python
def bubble_sort (arr):
    for j in range (len(arr)):
        for i in range (len(arr) - 1 - j):
            if arr[i] > arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
    return arr

arr = list(map(int,input().split()))
print (bubble_sort(arr))
```

### Bài 2: Sắp xếp chọn (Selection Sort) - Dễ
**Đề bài:** Viết hàm `selection_sort(arr)` hoạt động bằng cách liên tục tìm phần tử nhỏ nhất từ phần chưa sắp xếp và đưa nó lên đầu.
* **Input:** Một danh sách số nguyên `arr`.
* **Output:** Danh sách đã được sắp xếp tăng dần.
* **Ví dụ:** `selection_sort([64, 25, 12, 22, 11])` -> `[11, 12, 22, 25, 64]`.

Chia mảng thành hai phần: phần đã sắp xếp và phần chưa sắp xếp. Thuật toán tìm kiếm phần tử nhỏ nhất từ phần chưa sắp xếp và hoán đổi nó với phần tử đầu tiên của vùng này.

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        m = i
        for j in range(i + 1, n):
            if arr[j] < arr[m]:
                m = j
        arr[i], arr[m] = arr[m], arr[i]
    return arr

arr = list(map(int, input().split()))
print(selection_sort(arr))
```

### Bài 3: Sắp xếp chèn (Insertion Sort) - Vừa phải
**Đề bài:** Viết hàm `insertion_sort(arr)` thực hiện sắp xếp chèn bằng cách xây dựng dần mảng đã sắp xếp ở phía đầu.
* **Input:** Một danh sách số nguyên `arr`.
* **Output:** Danh sách đã được sắp xếp tăng dần.
* **Ví dụ:** `insertion_sort([12, 11, 13, 5, 6])` -> `[5, 6, 11, 12, 13]`.

Xây dựng mảng đã sắp xếp bằng cách lấy từng phần tử từ mảng chưa sắp xếp và "chèn" nó vào vị trí chính xác trong phần đã sắp xếp.

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        k = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > k:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = k
    return arr
```

### Bài 4: Sắp xếp nhanh (Quick Sort) - Vừa phải
**Đề bài:** Viết hàm `quick_sort_algo(arr)` sử dụng thuật toán Quick Sort để sắp xếp mảng tăng dần.
* **Input:** Một danh sách số nguyên `arr`.
* **Output:** Danh sách đã được sắp xếp tăng dần.
* **Ví dụ:** `quick_sort_algo([10, 7, 8, 9, 1, 5])` -> `[1, 5, 7, 8, 9, 10]`.

```python
def quick_sort_algo(arr):
    if len(arr) <= 1:
        return arr
    else:
        mid = arr[len(arr)//2]
        left = [i for i in arr if i < mid]
        right = [i for i in arr if i > mid]
        return quick_sort_algo(left) + [mid] + quick_sort_algo(right)       

arr = list(map(int, input().split()))
print (quick_sort_algo(arr))
```

### Bài 5: Sắp xếp trộn (Merge Sort) - Vừa phải
**Đề bài:** Viết hàm `merge_sort_algo(arr)` chia mảng thành hai nửa, sắp xếp đệ quy từng nửa rồi trộn (merge) chúng lại với nhau.
* **Input:** Một danh sách số nguyên `arr`.
* **Output:** Danh sách đã được sắp xếp tăng dần.
* **Ví dụ:** `merge_sort_algo([38, 27, 43, 3, 9, 82, 10])` -> `[3, 9, 10, 27, 38, 43, 82]`.

Chia đôi mảng liên tục cho đến khi mỗi mảng con chỉ còn đúng 1 phần tử (mảng 1 phần tử luôn coi là đã sắp xếp). Sau đó, tiến hành trộn (merge) các mảng con lại với nhau theo thứ tự tăng dần.

```python
def merge_sort(arr):
    n = len(arr)
    if n <= 1:
        return arr
    
    m = n // 2
    l = merge_sort(arr[:m])
    r = merge_sort(arr[m:])
    
    return merge(l, r)

def merge(l, r):
    res = []
    i = j = 0
    
    while i < len(l) and j < len(r):
        if l[i] < r[j]:
            res.append(l[i])
            i += 1
        else:
            res.append(r[j])
            j += 1
            
    res.extend(l[i:])
    res.extend(r[j:])
    return res

arr = list(map(int, input().split()))
print(merge_sort(arr))
```
