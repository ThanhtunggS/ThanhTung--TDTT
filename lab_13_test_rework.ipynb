Viết chương trình để mã hóa một số nguyên dương có 4 chữ số để truyền tin bảo mật. Quy tắc mã hóa như sau:
Thay thế mỗi chữ số bằng kết quả của phép tính: (chữ số + 7) % 10.
Đổi chỗ chữ số thứ nhất cho chữ số thứ ba.
Đổi chỗ chữ số thứ hai cho chữ số thứ tư.
In ra số đã mã hóa.
Lưu ý: Phải đảm bảo in đủ 4 chữ số (ví dụ nếu kết quả là 0189, hãy in 0189 thay vì 189).

```python
def encryption (n):
  h = [int(i) for i in n]
  k = []
  for i in h:
    i = (i + 7)%10
    k.append(i)
  a = k[0]
  b = k[1]
  c = k[2]
  d = k[3]
  k[0] = c
  k[1] = d
  k[2] = a
  k[3] = b
  return k

n = input()
print (*encryption(n), sep ='')
```

Viết chương trình nén chuỗi bằng cách đếm số lượng ký tự liên tiếp giống nhau.
Ví dụ: AAABBB -> 3A3B.
Lưu ý: Chương trình cần phân biệt được chữ hoa, chữ thường, số và khoảng trắng.

```python
def run_length_encoding(s):
  count = 1
  new_lst = []
  for i in range (len(s)):
    if i <= len(s) - 2:
      if s[i] == s[i + 1]:
        count += 1  
      else:
        new_lst.append(count)
        new_lst.append(s[i])
        count = 1
    if i == len(s) - 1:
      new_lst.append(count)
      new_lst.append(s[i])
  return new_lst
s = str(input())
print (*run_length_encoding(s), sep = '')
```
