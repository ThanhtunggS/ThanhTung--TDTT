thức kiểm tra xem có phải gốc tọa dộ không, có nằm trên trục 
tung không, có nằm trên trục hoành không, tính khoảng cách từ 
điểm đó tới một điểm khác 
 
Đầu vào:  
Nhập vào tọa độ một điểm M bất kỳ 
 
Đầu ra: 
Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục 
tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ 
là bao nhiêu (lấy 2 chữ số phần thập phân)

```python
import math

class Point:
    def __init__(self, x, y):
      self.x = x
      self.y = y

    def is_origin(self):
      return (self.x == 0 and self.y == 0)

    def is_on_y_axis(self):
        return (self.x != 0 and self.y == 0)

    def is_on_x_axis(self):
        return (self.x == 0 and self.y != 0)

    def distance_to(self, other):
        return (self.x ** 2 + self. y ** 2)**0.5

x, y = map(float, input().split())

M = Point(x, y)
O = Point(0, 0)

if M.is_origin():
    print("M là gốc tọa đọ")
elif M.is_on_y_axis():
    print("M trên trục hoành")
elif M.is_on_x_axis():
    print("M trên trục tung")
else:
    print("M ko nằm trên trục nào")

dist = M.distance_to(O)
print(f"{dist:.2f}")
```

Xây dựng lớp Calculator (máy tính bỏ túi) với các phương thức 
để thực hiện các phép toán 1. add (cộng), 2. subtract(trừ), 3. 
product (nhân) và 4. divide(chia), 5. power(hàm mũ), 6. mod 
(lấy phần dư) và phương thức 7. set_numbers(cập nhật 2 số 
mới) nếu người dùng muốn tính toán trên 2 số khác. 
Khởi tạo lớp là hai thuộc tính gồm 2 số cần tính, Mỗi phương 
thức thực hiện một phép tính và trả về kết quả tương ứng 
Tạo menu và cho người dùng tính toán tới khi gõ "yes", "YES" 
hoặc "Yes" cho câu hỏi "Bạn có muốn thoát?" mới dừng.

```python
import math

def uoc_chung_lon_nhat (a, b):
  return math.gcd(a,b) 

class Fraction:
  def __init__ (self, num, den):
    if den < 0:
      num = -num
      den = -den
    c = uoc_chung_lon_nhat(num, den)
    self.n = num//c
    self.d = den//c

  def __str__(self):
    return f"{self.n}/{self.d}"

  def add(self, o):
    return Fraction(self.n * o.d + o.n * self.d, self.d * o.d)

  def sub(self, o):
    return Fraction(self.n * o.d - o.n * self.d, self.d * o.d)  

  def mul(self, o):
    return Fraction(self.n * o.n, self.d * o.d)

  def div(self, o):
    return Fraction(self.n * o.d, self.d * o.n)

a,b,op,c,d = input().split()
f1 = Fraction(int(a), int(b))
f2 = Fraction(int(c), int(d))

if op == '+': print (f1.add(f2))
elif op == '-': print (f1.sub(f2))
elif op == '*': print (f1.mul(f2))
elif op == '/': print (f1.div(f2))
```

Khi bạn truyền một đối tượng vào hàm print(), Python sẽ tự động tìm kiếm và kích hoạt phương thức __str__ được định nghĩa bên trong lớp của đối tượng đó để lấy về một chuỗi ký tự (string) trước khi xuất ra màn hình.

Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản 
phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi     
sản phẩm là một từ điển có name (tên) và price(giá).  
Với các phương thức sau: 
1. Khởi tạo với danh sách rỗng 
2. Thêm vào giỏ hàng một sản phẩm mới 
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa) 
4. Kiểm tra xem giỏ hàng có rỗng không 
5. Tính tổng tiền giỏ hàng

Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có 
muốn thoát khỏi chương trình không?

```python
class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, name, price):
        item = {"name": name, "price": price}
        self.items.append(item)

    def remove_item(self, name):
        for item in self.items:
            if item["name"] == name:
                self.items.remove(item)
                return True
        return False

    def is_empty(self):
        return len(self.items) == 0

    def get_total(self):
        total = 0
        for item in self.items:
            total += item["price"]
        return total

    def show_cart(self):
        if self.is_empty():
            print("Gio hang trong.")
        else:
            for item in self.items:
                print(f"- {item['name']}: {item['price']}")

    def clear_cart(self):
        self.items.clear()


cart = ShoppingCart()

while True:
    choice = input()

    if choice == "2":
        name = input()
        price = float(input())
        cart.add_item(name, price)
        
    elif choice == "3":
        name = input()
        cart.remove_item(name)
            
    elif choice == "4":
        if cart.is_empty():
            print("rong.")
        else:
            print("ko rong.")

    elif choice == "5":
        print(cart.get_total())
    elif choice == "6":
        cart.show_cart()
        
    elif choice == "7":
        cart.clear_cart()
        print("Da xoa toan bo gio hang.")
        
    else:
        print("Lua chon khong hop le.")

    exit_choice = input("Co muon thoat khoi chuong trinh khong? (yes/no): ")
    if exit_choice.lower() == "yes":
        break
```

Xây dựng lớp Rectangle(w, h) với các phương thức: 
 
area() trả về diện tích. 
 
perimeter() trả về chu vi. 
 
scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h), 
không trả về gì. 
 
Input 
w h k (số thực dương) 
 
Output 
area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân).

```python
class Rectangle:
    def __init__ (self, w, h):
        self.width = w
        self.height = h
    def area (self):
        return self.width * self.height
    def perimeter (self):
        return 2*(self.width + self.height)
    def scale (self, k):
        self.width *= k
        self.height *= k

w, h, k = map(float, input().split())

rect = Rectangle(w, h)
rect.scale(k)

print(round(rect.area(), 2), round(rect.perimeter(), 2))
```

```python
class BankAccount:
    def __init__ (self, balance):
        self.b = balance

    def deposit (self, x):
        self.b += x
    def withdraw (self, y):
        self.b -= y
    def __str__ (self):
        return (f"owner:{self.b}")    

b = int(input())
n = int(input())
bankac = BankAccount(b)

for _ in range (n):
    x = int(input())
    k = input()
    if k == "deposit":
        bankac.deposit (x)
    if k == "withdraw":
        if x > bankac:
            print ("u want a loan?")
            break
        else:
            bankac.withdraw (x)

print (bankac)

```

Lớp Student(name) với: 
 
add_score(subject, score) 
 
avg() trả về điểm trung bình. 
 
rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8, 
"Average" nếu 5–<6.5, ngược lại "Poor". 
 
Input 
Dòng 1: name 
Dòng 2: n 
n dòng tiếp: subject score (score là số thực 0–10) 
 
Output 
name avg rank (avg 2 chữ số).

```python
class Student:
    def __init__ (self, name):
        self.n = name
        self.score = []
    def add_score (self, subject, score):
        name = {"subject": subject, "score": score}
        self.score.append(name)
    def average (self):
        tong = sum(i["score"] for i in self.score)
        self.trungbinh = (tong/len(self.score))  #gọi giá trị: Ko cần ()
        return self.trungbinh
    def rank (self):
        if self.trungbinh >= 8: return "Excellent"
        elif 6.5 <= self.trungbinh <= 8: return "Good"
        elif 5 <= self.trungbinh <= 6.5: return "Poor"
    def __str__ (self):
        return f"name: {self.n}, avg: {self.average():.2f}, rank: {self.rank()}" #Gọi hàm: Cần ()

name = str(input())
n = int(input())
stu = Student(name)
for _ in range (n):
    in4 = input().split()
    subject = in4[0]
    point = int(in4[1])
    stu.add_score(subject, point)
print (stu)
```

Yêu cầu 
 
Lớp Book(title, author, year). 
 
Lớp Library() với: 
 
add(book) 
 
count_by_author(name) 
 
find_by_year(y) trả về số sách xuất bản năm y. 
 
Input 
Dòng 1: n (số thao tác) 
n dòng: 
 
ADD title;author;year 
 
COUNT author (đếm theo tác giả) 
 
COUNTYEAR year (đếm theo năm) 
 
Output 
Mỗi lệnh COUNT/COUNTYEAR in một dòng kết quả

```python
class Book:
    def __init__(self, title: str, author: str, year: int):
        self.title = title
        self.author = author
        self.year = year

class Library:
    def __init__ (self):
        self.books = []

    def add (self, Book):
        self.books.append(Book)

    def count_by_author (self, name) -> int: #Các hàm phải gọi thêm biến sau self (VD:(self, name)) là để thêm giá trị từ ngoài vào (name)
        count = 0
        for i in self.books:
            if i.author == name:   #i.author để lấy tên tác giả trong các cuốn sách
                count += 1
        return count

    def find_by_year (self, year) -> int: # -> int: là để đảm bảo hàm luôn trả về kết quả là int
        count1 = 0
        for book in self.books:
            if book.year == year:   #Ở đây ta phải dùng year đứng đơn lẻ (ko phải self.year) do đang làm việc với giá trị nhập ngoài vào
                count1 += 1
        return count1 

n = int(input())
my_lib = Library() #Phải khởi tạo đối tượng cụ thể mới dùng hàm được
for _ in range (n):
    k = list(input().split(" ", 1)) #Tách ra, phần " " là đại diện cho lệnh ADD hoặc..., còn phần 1 ở phía sau là cả cụm thông tin đầu vào
    if k[0] == "ADD":
        info = k[1].split(";")
        title = info[0]
        author = info[1]
        year = int(info[2])
        book1 = Book (title, author, year)
        my_lib.add(book1)
    elif k[0] == "COUNT":
        name = k[1]
        print (my_lib.count_by_author(name))  
    elif k[0] == "COUNTYEAR":
        year = int(k[1])  
        print (my_lib.find_by_year(year))
```

#### W8A14  Yêu cầu 
Xây dựng lớp Animal với phương thức speak(). 
Xây dựng hai lớp kế thừa Dog và Cat, mỗi lớp ghi đè speak() 
để trả về âm thanh tương ứng: 
- Dog -> Woof 
- Cat -> Meow 
 
Input 
Một dòng chứa tên lớp: Dog hoặc Cat. 
Dog  Woof 
 
Output 
In ra kết quả của speak()

```python
# Trước hết, thiết kế class cha: Animal. Nó định nghĩa con vật nào cx có khả năng kêu
# Thứ hai, Class Animal ở đây thậm chí ko cần hàm __init__ nhập tên, vì bản thân tên của lớp Dog hay Cat đã xác định nó là con gì rồi
class Animal:
    def speak(self):
        return ""

class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

name = input()
if name == "Dog":
    animal = Dog() #Cần chuyển sang hàm
if name == "Cat":
    animal = Cat()

print (animal.speak())

    
```

Đề bài bắt buộc bạn tạo lớp cha Animal và hai lớp con Dog, Cat kế thừa từ Animal để bạn làm quen với cách:

Lớp con kế thừa các thuộc tính/phương thức từ lớp cha.
Lớp con ghi đè (override) lại phương thức speak() của lớp cha để định nghĩa hành vi riêng biệt cho mình.
