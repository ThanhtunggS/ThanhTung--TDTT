#### W6A2 
Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu 
tới vị trí i), giữ nguyên thứ tự 

```python
def tinh_tong_tich_kuy (n):
    tap_so = [0, 1]
    for i in range (1, n):
        tong = sum(tap_so[:i+1])
        tap_so.append (tong)
    return tap_so
n = int(input())
print (*tinh_tong_tich_kuy(n), sep = ", ")
```

#### W6A4

Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo 
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key 
lặp lại, thêm vào list của key đó 

```python
a = input().split()
d = {}

for i in range (0, len(a), 2):
    k = a[i]
    v = int(a[i + 1])
    if k not in d:
        d[k] = [v]
    else:
        d[k].append(v)
print (d)        
```

#### W6A5 

Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 
'zeros': ...} (đếm số >0, <0, =0) 

```python
list_so = list(map(int,input().split()))
k = ['positives', 'negatives', 'zeros']
pos = []
neg = []
zero = []
d = dict.fromkeys(k)
for i in list_so:
    if i > 0:
        pos.append(i)
    if i < 0:
        neg.append(i)
    if i == 0:
        zero.append(i)
d["positives"] = pos
d["negatives"] = neg
d["zeros"] = zero

print (d)        
```

```python
def reverse_tuple(n):
    listn = list(n)
    listn.reverse()
    return tuple(listn)
s = input()
a = s.split()
n = tuple(a)
print (n[0])
print (n[-1])
print (reverse_tuple(n))
```
