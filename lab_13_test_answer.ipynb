Thay thế mỗi chữ số bằng kết quả của phép tính: (chữ số + 7) % 10.
Đổi chỗ chữ số thứ nhất cho chữ số thứ ba.
Đổi chỗ chữ số thứ hai cho chữ số thứ tư.
In ra số đã mã hóa.
Lưu ý: Phải đảm bảo in đủ 4 chữ số (ví dụ nếu kết quả là 0189, hãy in 0189 thay vì 189).

```python
def encryption (n):
  h = [int(i) for i in n]
  k = []
  for i in h:
    i = (i + 7)%10
    k.append(i)
  a = k[0]
  b = k[1]
  c = k[2]
  d = k[3]
  k[0] = c
  k[1] = d
  k[2] = a
  k[3] = b
  return k

n = input()
print (*encryption(n), sep ='')
```

```python
def is_leap_year(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def days_in_month(month, year):
    if month in (1, 3, 5, 7, 8, 10, 12):
        return 31
    if month in (4, 6, 9, 11):
        return 30
    if month == 2:
        return 29 if is_leap_year(year) else 28
    return 0

def main():
    try:
        inp = input().strip()
        inp = inp.replace(',', ' ')
        parts = inp.split()
        if len(parts) != 3:
            print("Invalid Date")
            return
        
        day = int(parts[0])
        month = int(parts[1])
        year = int(parts[2])
        
        if year < 1 or month < 1 or month > 12 or day < 1 or day > days_in_month(month, year):
            print("Invalid Date")
            return
            
        next_d, next_m, next_y = day + 1, month, year
        if next_d > days_in_month(month, year):
            next_d = 1
            next_m += 1
            if next_m > 12:
                next_m = 1
                next_y += 1
                
        print(f"{next_d}/{next_m}/{next_y}")
    except ValueError:
        print("Invalid Date")

main()
```

Viết chương trình nén chuỗi bằng cách đếm số lượng ký tự liên tiếp giống nhau.
Ví dụ: AAABBB -> 3A3B.
Lưu ý: Chương trình cần phân biệt được chữ hoa, chữ thường, số và khoảng trắng.

```python
def run_length_encoding(s):
  count = 1
  new_lst = []
  for i in range (len(s)):
    if i <= len(s) - 2:
      if s[i] == s[i + 1]:
        count += 1  
      else:
        new_lst.append(count)
        new_lst.append(s[i])
        count = 1
    if i == len(s) - 1:
      new_lst.append(count)
      new_lst.append(s[i])
  return new_lst
s = str(input())
print (*run_length_encoding(s), sep = '')
```

Nhập vào 1 xâu kí tự s và 1 khóa k. In ra mã hóa Caesar của xâu ký tự này với khóa được nhập.

Mã hóa Caesar là một dạng mật mã thay thế, trong đó mỗi ký tự trên văn bản thô sẽ được thay bằng một ký tự khác, có vị trí cách nó một khoảng xác định trong bảng chữ cái. Ví dụ với độ dịch chuyển là 3, A sẽ trở thành D, B sẽ trở thành E, Z sẽ trở thành C, 9 sẽ trở thành 2 v.v... Những ký tự nào không thuộc bảng chữ cái được giữ nguyên.

VD1:
Input:
abcx
4
Output:
efgb

```python
def caesar_cipher(s, k):
    result = []
    for char in s:
        if 'a' <= char <= 'z':
            result.append(chr((ord(char) - ord('a') + k) % 26 + ord('a')))
        elif 'A' <= char <= 'Z':
            result.append(chr((ord(char) - ord('A') + k) % 26 + ord('A')))
        elif '0' <= char <= '9':
            result.append(chr((ord(char) - ord('0') + k) % 10 + ord('0')))
        else:
            result.append(char)
    return "".join(result)

s = input()
k = int(input())
print(caesar_cipher(s, k))
```

Viết chương trình yêu cầu người dùng nhập vào một chuỗi ngoặc chỉ gồm các ký tự "(", ")", "[", "]", "{", "}".
Kiểm tra chuỗi có hợp lệ hay không. Nếu hợp lệ in ra "Valid", nếu không in ra "Invalid". Chuỗi hợp lệ nếu:
- Ngoặc được đóng đúng loại
- Thứ tự đóng ngoặc đúng (dạng LIFO - Last In First Out)
Ví dụ:
()[]{} -> Valid
(] -> Invalid
([)] -> Invalid
([]) -> Valid

```python
def is_valid_brackets(s):
    stack = []
    mapping = {")": "(", "]": "[", "}": "{"}
    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping.keys():
            if not stack or stack[-1] != mapping[char]:
                return "Invalid"
            stack.pop()
        else:
            return "Invalid"
    return "Valid" if not stack else "Invalid"

s = input().strip()
print(is_valid_brackets(s))
```

Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. Mẫu hàm: def binary_search(arr, left, right, target), với arr là danh sách (list) đầu vào, left: là chỉ số trái, right: chỉ số phải, target: giá trị cần tìm kiếm.

```python
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
```

```python
def check_digit_sums(n):
    s = str(n)
    even_sum = 0
    odd_sum = 0
    for i in range(len(s)):
        if i % 2 == 0:
            even_sum += int(s[i])
        else:
            odd_sum += int(s[i])
    return even_sum == odd_sum
```

Viết hàm kiểm tra xem một số nguyên dương n có phải là số thiếu hay không, biết rằng số thiểu là số có tổng các ước số thực sự nhỏ hơn chính số đó.

Ví dụ: Số 8 là số thiếu vì tổng các ước 1 + 2 + 4 = 7 < 8.

```python
def is_deficient_number(n):
    if n <= 1:
        return True
    divisor_sum = 0
    for i in range(1, n):
        if n % i == 0:
            divisor_sum += i
    return divisor_sum < n
```

Viết hàm thực hiện thuật toán bubble sort hoặc thuật toán Selection Sort để sắp xếp danh sách theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort); Sắp xếp chọn (Selection Sort)

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
```

Cho một mảng số nguyên không âm arr. Gọi mảng đó là "mảng đẹp" nếu mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng arr có phải là mảng đẹp hay không. Nếu không hãy xóa đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

1 2 2 1 1 3 3

```python
def min_deletions_to_beautiful(arr):
    counts = {}
    for x in arr:
        counts[x] = counts.get(x, 0) + 1
        
    deletions = 0
    for x, count in counts.items():
        if count >= x:
            deletions += min(count, count - x)
        else:
            deletions += count
    return deletions

arr = [int(i) for i in input().split()]
print(min_deletions_to_beautiful(arr))
```

Cho hai mảng đã sắp xếp (tăng dần hoặc giảm dần). Hãy trộn hai mảng này thành một mảng cùng thứ tự sắp xếp mà không được dùng bất cứ một thuật toán sắp xếp nào.
VD:
input output
1 2 3 4 5 6 70
0 13 18 20 0 1 2 3 4 5 6 13 18 20 70
5 4 3 2
10 9 8 1 0 10 9 8 5 4 3 2 1 0

```python
def merge_increasing(arr1, arr2):
    i, j = 0, 0
    res = []
    while i < len(arr1) and j < len(arr2):
        if arr1[i] <= arr2[j]:
            res.append(arr1[i])
            i += 1
        else:
            res.append(arr2[j])
            j += 1
    res.extend(arr1[i:])
    res.extend(arr2[j:])
    return res

def merge_decreasing(arr1, arr2):
    i, j = 0, 0
    res = []
    while i < len(arr1) and j < len(arr2):
        if arr1[i] >= arr2[j]:
            res.append(arr1[i])
            i += 1
        else:
            res.append(arr2[j])
            j += 1
    res.extend(arr1[i:])
    res.extend(arr2[j:])
    return res

def main():
    arr1 = [int(x) for x in input().split()]
    arr2 = [int(x) for x in input().split()]
    
    if not arr1:
        print(*arr2)
        return
    if not arr2:
        print(*arr1)
        return
        
    is_increasing = True
    if len(arr1) > 1:
        is_increasing = arr1[0] <= arr1[-1]
    elif len(arr2) > 1:
        is_increasing = arr2[0] <= arr2[-1]
        
    if is_increasing:
        res = merge_increasing(arr1, arr2)
    else:
        res = merge_decreasing(arr1, arr2)
        
    print(*res)

main()
```

Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không và phương thức tính ngày kế tiếp của ngày tháng năm nào đó nếu nó là hợp lệ

Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó.

Đầu vào
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng dd/mm/yyyy
.
Đầu ra
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID

```python
class Date:
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year

    def is_leap_year(self):
        y = self.year
        return (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)

    def days_in_month(self):
        m, y = self.month, self.year
        if m in (1, 3, 5, 7, 8, 10, 12):
            return 31
        if m in (4, 6, 9, 11):
            return 30
        if m == 2:
            return 29 if self.is_leap_year() else 28
        return 0

    def is_valid(self):
        if self.year < 1 or self.month < 1 or self.month > 12 or self.day < 1:
            return False
        return self.day <= self.days_in_month()

    def get_next_date(self):
        if not self.is_valid():
            return None
        
        next_d = self.day + 1
        next_m = self.month
        next_y = self.year
        
        if next_d > self.days_in_month():
            next_d = 1
            next_m += 1
            if next_m > 12:
                next_m = 1
                next_y += 1
                
        return Date(next_d, next_m, next_y)

    def __str__(self):
        return f"{self.day:02d}/{self.month:02d}/{self.year:04d}"


def main():
    try:
        inp = input().strip()
        parts = inp.split('/')
        if len(parts) != 3:
            print("INVALID")
        else:
            d = int(parts[0])
            m = int(parts[1])
            y = int(parts[2])
            
            date = Date(d, m, y)
            if not date.is_valid():
                print("INVALID")
            else:
                next_date = date.get_next_date()
                print(next_date)
    except ValueError:
        print("INVALID")

main()
```

Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao (kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện tích toàn phần) và thể tích của hình trụ
Đầu vào
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và chiều cao của hình trụ, cách nhau bởi một dấu cách.

Đầu ra
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử dụng lại những hàm vừa viết, π=3.14).

(Kết quả làm tròn đến 2 chữ số sau dấu phẩy).

```python
class Cylinder:
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def surface_area(self):
        pi = 3.14
        return 2 * pi * self.radius * (self.height + self.radius)

    def volume(self):
        pi = 3.14
        return pi * (self.radius ** 2) * self.height

def main():
    try:
        parts = input().strip().split()
        r = float(parts[0])
        h = float(parts[1])
        
        cyl = Cylinder(r, h)
        area = round(cyl.surface_area(), 2)
        vol = round(cyl.volume(), 2)
        print(f"{area:.2f} {vol:.2f}")
    except (ValueError, IndexError):
        pass

main()
```

Mô tả: Bạn cần thiết kế một lớp (class) có tên là BankAccount để mô phỏng hoạt động của một tài khoản ngân hàng. Lớp này cần đảm bảo các yêu cầu kỹ thuật sau:
1. Hàm khởi tạo (__init__):
Nhận vào owner (tên chủ tài khoản - chuỗi ký tự) và balance (số dư ban đầu - số thực).
2. Phương thức deposit(amount):
Nạp số tiền amount vào tài khoản.Tham số amount là một số thực dương (amount > 0).
3. Phương thức withdraw(amount):
Rút số tiền amount từ tài khoản.Tham số amount là một số thực dương (amount > 0).
Điều kiện: Nếu amount lớn hơn số dư hiện tại (balance), giao dịch bị hủy bỏ (không trừ tiền, không báo lỗi).
4. Phương thức __str__():
Trả về chuỗi thông tin theo định dạng: Owner: Balance (trong đó Balance làm tròn đến 2 chữ số thập phân).

Dữ liệu đầu vào (Input):
Dòng 1: Chứa một chuỗi ký tự S (tên chủ tài khoản, không chứa khoảng trắng) và một số thực B (số dư ban đầu), ngăn cách bởi dấu cách.
Dòng 2: Chứa số nguyên N (1 <= N <= 100) biểu thị số lượng giao dịch.
N dòng tiếp theo: Mỗi dòng chứa một chuỗi ký tự CMD (lệnh) và một số thực X. CMD có thể là "DEPOSIT" hoặc "WITHDRAW".
Dữ liệu đầu ra (Output):In ra số dư cuối cùng của tài khoản, làm tròn đến 2 chữ số thập phân.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = float(balance)

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount

    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount

    def __str__(self):
        return f"{self.owner}: {self.balance:.2f}"

def main():
    try:
        parts = input().strip().split()
        owner = parts[0]
        balance = float(parts[1])
        
        account = BankAccount(owner, balance)
        
        n = int(input())
        for _ in range(n):
            cmd_parts = input().strip().split()
            cmd = cmd_parts[0]
            x = float(cmd_parts[1])
            
            if cmd == "DEPOSIT":
                account.deposit(x)
            elif cmd == "WITHDRAW":
                account.withdraw(x)
                
        print(f"{account.balance:.2f}")
    except (ValueError, IndexError):
        pass

main()
```

```python
def gcd(x, y):
    while y:
        x, y = y, x % y
    return abs(x)

class Fraction:
    def __init__(self, num, den):
        if den == 0:
            raise ValueError("Denominator cannot be zero.")
        if den < 0:
            num = -num
            den = -den
        g = gcd(num, den)
        self.num = num // g
        self.den = den // g

    def add(self, other):
        n = self.num * other.den + other.num * self.den
        d = self.den * other.den
        return Fraction(n, d)

    def sub(self, other):
        n = self.num * other.den - other.num * self.den
        d = self.den * other.den
        return Fraction(n, d)

    def mul(self, other):
        n = self.num * other.num
        d = self.den * other.den
        return Fraction(n, d)

    def div(self, other):
        n = self.num * other.den
        d = self.den * other.num
        return Fraction(n, d)

    def __str__(self):
        return f"{self.num}/{self.den}"

def main():
    try:
        parts = input().strip().split()
        if len(parts) == 5:
            a = int(parts[0])
            b = int(parts[1])
            op = parts[2]
            c = int(parts[3])
            d = int(parts[4])
            
            f1 = Fraction(a, b)
            f2 = Fraction(c, d)
            
            if op == '+':
                res = f1.add(f2)
            elif op == '-':
                res = f1.sub(f2)
            elif op == '*':
                res = f1.mul(f2)
            elif op == '/':
                res = f1.div(f2)
                
            print(res)
    except Exception:
        pass

main()
```

Bạn hãy thiết kế một lớp đối tượng (class) có tên là Student để quản lý điểm số của sinh viên. Lớp này cần đảm bảo các yêu cầu:
1. Hàm khởi tạo (init):
Nhận vào tham số name (tên sinh viên).
Khởi tạo danh sách điểm số rỗng.
2. add_score(subject, score):
Thêm điểm số của một môn học vào danh sách.
subject: Tên môn học (chuỗi ký tự).
score: Điểm số (số thực từ 0 đến 10).
3. avg():
Tính và trả về điểm trung bình cộng của tất cả các môn đã nhập.
Kết quả làm tròn đến 2 chữ số thập phân.
4. rank():
Trả về xếp loại học lực dựa trên điểm trung bình (avg) theo quy tắc sau:
"Excellent" nếu avg >= 8.0
"Good" nếu 6.5 <= avg < 8.0
"Average" nếu 5.0 <= avg < 6.5
"Poor" nếu avg < 5.0
DỮ LIỆU ĐẦU VÀO (INPUT)
Dòng 1: Tên sinh viên (chuỗi ký tự, không chứa khoảng trắng).
Dòng 2: Số nguyên n (số lượng môn học cần nhập).
n dòng tiếp theo: Mỗi dòng chứa tên môn học và điểm số, ngăn cách bởi dấu cách.
DỮ LIỆU ĐẦU RA (OUTPUT)
In ra một dòng duy nhất theo định dạng: name avg rank
Trong đó avg hiển thị đúng 2 chữ số thập phân.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, subject, score):
        self.scores.append((subject, float(score)))

    def avg(self):
        if not self.scores:
            return 0.0
        total = sum(item[1] for item in self.scores)
        return round(total / len(self.scores), 2)

    def rank(self):
        a = self.avg()
        if a >= 8.0:
            return "Excellent"
        elif a >= 6.5:
            return "Good"
        elif a >= 5.0:
            return "Average"
        else:
            return "Poor"

def main():
    try:
        name = input().strip()
        n = int(input())
        student = Student(name)
        for _ in range(n):
            parts = input().strip().split()
            subject = parts[0]
            score = float(parts[1])
            student.add_score(subject, score)
            
        print(f"{student.name} {student.avg():.2f} {student.rank()}")
    except (ValueError, IndexError):
        pass

main()
```

Giả sử bạn là nhân viên của một công ty cây xanh, được giao cho chăm sóc một luống đất. Luống đất đó được chia thành nhiều ô đất nhỏ có kích cỡ bằng nhau. Trước khi được giao cho bạn chăm sóc, trên luống hoa đã được trồng sẵn một số cây xanh tán rộng. Một ngày đầu xuân, công ty gửi đến cho bạn một số cây hoa và yêu cầu bạn trồng vào những vị trí còn trống trong luống hoa. Đây là những cây hoa hướng sáng và chúng không thể sống được dưới bóng râm của cây khác, đặc biệt là cây tán rộng. Vì vậy, bạn không thể trồng chúng tại ô bên cạnh những ô đã được trồng cây xanh tán rộng. Và nếu trồng những cây hoa bên cạnh nhau, chúng sẽ tranh giành phân bón và nước và sau cùng, cả hai cây sẽ cùng chết.

Viết hàm canPlaceFlowers(flowerbed, k) kiểm tra xem có thể trồng hết số hoa mới được cấp vào trong luống đất của bạn hay không.

Hàm nhận đầu vào là một danh sách (list) flowerbed, đánh dấu những vị trí trong luống đất đã được trồng cây xanh trước đó (giá trị 1 ứng với vị trí cây đã được trồng và 0 ứng với vị trí đất trống) và số lượng hoa được phát thêm k (int). Hàm trả về True nếu có thể trồng hết toàn bộ k cây hoa vào luống đất và False trong trường hợp ngược lại.

```python
def canPlaceFlowers(flowerbed, k):
    count = 0
    n = len(flowerbed)
    f = flowerbed.copy()
    
    for i in range(n):
        if f[i] == 0:
            left_empty = (i == 0) or (f[i - 1] == 0)
            right_empty = (i == n - 1) or (f[i + 1] == 0)
            
            if left_empty and right_empty:
                f[i] = 1
                count += 1
                if count >= k:
                    return True
                    
    return count >= k
```
