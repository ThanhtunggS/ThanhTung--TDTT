```python
def gt(n):
    if n <= 1:
        return 1
    return n * gt(n - 1)

print (gt(5))
```

```python
#Tinh fibonaci thu n:0, 1, 1, 2, 3, 5, 8
def fibo(n):
    if n <= 1:
        return 0
    if n == 2:
        return 1
    if n >= 3:
        return fibo(n- 1) + fibo(n- 2)
    
print (fibo(5))    
```
