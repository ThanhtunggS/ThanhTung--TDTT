Cho một dãy số nguyên a gồm N phần tử và một số nguyên dương M (M < N).
Ta có thể chia dãy a thành các mảng con liên tiếp, mỗi mảng con có độ dài đúng bằng M. Tổng cộng sẽ có N - M + 1 mảng con như vậy.
Nhiệm vụ của bạn là tìm giá trị nhỏ nhất trong từng mảng con đó và in ra theo thứ tự từ trái sang phải.

Dữ liệu vào (Input):
Dòng 1: Chứa 2 số nguyên N và M cách nhau bởi dấu cách.
Dòng 2: Chứa N số nguyên của mảng a, cách nhau bởi dấu cách.

In ra N - M + 1 số nguyên trên một dòng, mỗi số cách nhau một dấu cách. Số thứ i là giá trị nhỏ nhất của mảng con thứ i.

```python
def main():
    try:
        parts = input().strip().split()
        n = int(parts[0])
        m = int(parts[1])
        arr = [int(x) for x in input().strip().split()]
        
        results = []
        for i in range(n - m + 1):
            sub_arr = arr[i : i + m]
            results.append(min(sub_arr))
            
        print(*results)
    except (ValueError, IndexError):
        pass

main()
```

Cho dãy số nguyên a và số nguyên x. Tìm giá trị v trong a có |v - x| nhỏ nhất. Nếu có nhiều giá trị cùng khoảng cách, chọn giá trị nhỏ hơn.

```python
def find_closest(a, x):
    if not a:
        return None
    closest_v = a[0]
    min_diff = abs(a[0] - x)
    
    for v in a:
        diff = abs(v - x)
        if diff < min_diff:
            min_diff = diff
            closest_v = v
        elif diff == min_diff:
            if v < closest_v:
                closest_v = v
                
    return closest_v

try:
    arr = [int(i) for i in input().split()]
    target = int(input())
    print(find_closest(arr, target))
except Exception:
    pass
```

Cho list a và số nguyên k (1 ≤ k ≤ n). Trả về k phần tử lớn nhất theo thứ tự giảm dần

Lưu ý: Không dùng hàm sort có sẵn

```python
def get_k_largest(a, k):
    temp = a.copy()
    result = []
    for _ in range(k):
        if not temp:
            break
        max_val = temp[0]
        max_idx = 0
        for i in range(len(temp)):
            if temp[i] > max_val:
                max_val = temp[i]
                max_idx = i
        result.append(max_val)
        temp.pop(max_idx)
    return result

try:
    arr = [int(i) for i in input().split()]
    k = int(input())
    print(*get_k_largest(arr, k))
except Exception:
    pass
```

```python
def get_elements_appearing_twice(a):
    counts = {}
    for x in a:
        counts[x] = counts.get(x, 0) + 1
        
    result = []
    seen = set()
    for x in a:
        if counts[x] == 2 and x not in seen:
            result.append(x)
            seen.add(x)
    return result

try:
    arr = [int(i) for i in input().split()]
    print(get_elements_appearing_twice(arr))
except Exception:
    pass
```

Mô tả bài toán:Viết hàm đệ quy thực hiện tính x lũy thừa n.
Hàm cần xử lý được cả số mũ âm.
Tối ưu hóa thời gian (độ phức tạp O(log n)).

Input:
x: Số thực (float).
n: Số nguyên (integer).

Output:
Kết quả x^n (float).

```python
def power(x, n):
    if n == 0:
        return 1.0
    if n < 0:
        return 1.0 / power(x, -n)
    
    half = power(x, n // 2)
    if n % 2 == 0:
        return half * half
    else:
        return half * half * x

try:
    x = float(input())
    n = int(input())
    print(power(x, n))
except Exception:
    pass
```

Bạn đứng ở chân cầu thang gồm n bậc. Mỗi lần bạn có thể bước lên 1 bậc hoặc 2 bậc.
Hãy tính số cách khác nhau để lên đúng bậc n.

Input
Một số nguyên dương n (1 ≤ n ≤ 45)
Output
Một số nguyên: số cách leo lên bậc n
Ví dụ
Input: 2 -> Output: 2 ( (1+1), (2) )
Input: 3 -> Output: 3 ( (1+1+1), (1+2), (2+1) )

```python
def climb_stairs(n):
    if n <= 1:
        return 1
    a, b = 1, 2
    for _ in range(3, n + 1):
        a, b = b, a + b
    return b

try:
    n = int(input())
    print(climb_stairs(n))
except Exception:
    pass
```

Viết chương trình sử dụng khai triển Taylor để tính giá trị hàm mũ theo công thức sau:
\( e^x = 1 + x + {x^2}/{2!} + {x^3}/{3!} + {x^4}/{4!} + {x^5}/{5!} + ... \)
Sai số tới 0.001.
Với x là số nguyên dương. Kết quả làm tròn đến số thập phân thứ tư.
Gợi ý: sử dụng vòng lặp while và kiểm tra kiện (sum - sum_old > 0.001).

```python
def main():
    try:
        x = int(input())
        total = 1.0
        term = 1.0
        i = 1
        
        while True:
            old_total = total
            term = term * x / i
            total += term
            if total - old_total <= 0.001:
                break
            i += 1
            
        print(f"{total:.4f}")
    except Exception:
        pass

main()
```

Input

Dòng 2: n số nguyên — các phần tử của list1

Output
In ra:

```python
def is_subarray():
    try:
        n = int(input())
        list1 = [int(x) for x in input().strip().split()]
        m = int(input())
        list2 = [int(x) for x in input().strip().split()]
        
        if n == 0:
            print("YES")
            return
            
        for i in range(m - n + 1):
            if list2[i : i + n] == list1:
                print("YES")
                return
                
        print("NO")
    except (ValueError, IndexError):
        print("NO")

is_subarray()
```

Bạn được cho một dictionary D có:
key là chuỗi (không chứa khoảng trắng),
value là số nguyên.

Tiếp theo, bạn được cho một số nguyên k.
Hãy tạo một dictionary mới R chỉ gồm những cặp (key, value) trong D mà value > k.

Input
Dòng 1: số nguyên n — số lượng cặp key-value trong dictionary.
n dòng tiếp theo: mỗi dòng gồm key value
key là chuỗi (không có khoảng trắng)
value là số nguyên
Dòng cuối: số nguyên k
Output
Nếu R rỗng: in ra EMPTY
Ngược lại: in ra các cặp key value của R, mỗi cặp trên một dòng, theo thứ tự key tăng dần (từ điển).

```python
def filter_dict():
    try:
        n = int(input())
        D = {}
        for _ in range(n):
            parts = input().strip().split()
            key = parts[0]
            val = int(parts[1])
            D[key] = val
            
        k = int(input())
        
        R = {key: val for key, val in D.items() if val > k}
        
        if not R:
            print("EMPTY")
        else:
            for key in sorted(R.keys()):
                print(f"{key} {R[key]}")
    except (ValueError, IndexError):
        pass

filter_dict()
```

Ta định nghĩa một kiểu dữ liệu thời gian Time gồm 3 thành phần số nguyên hour, minute, second.

Khi tạo một class thời điểm Time(hour, minute, second), cần chuẩn hóa dữ liệu bằng cách:
Dồn phần giây vượt 60 sang phút
Dồn phần phút vượt 60 sang giờ
Nếu giờ vượt 24 thì quy về trong một ngày (mod 24)
Sau khi chuẩn hóa, so sánh hai thời điểm t1 và t2:
t1 sớm hơn t2
t2 sớm hơn t1
hoặc bằng nhau

Input
Gồm 2 dòng:
Dòng 1: hour minute second — thời điểm đến của Student 1
Dòng 2: hour minute second — thời điểm đến của Student 2
Output
In Student 1 nếu Student 1 đến sớm hơn
In Student 2 nếu Student 2 đến sớm hơn
Nếu đến cùng thời điểm, in Equal

```python
class Time:
    def __init__(self, hour, minute, second):
        minute += second // 60
        second %= 60
        hour += minute // 60
        minute %= 60
        hour %= 24
        
        self.hour = hour
        self.minute = minute
        self.second = second

    def total_seconds(self):
        return self.hour * 3600 + self.minute * 60 + self.second

def main():
    try:
        h1, m1, s1 = map(int, input().split())
        h2, m2, s2 = map(int, input().split())
        
        t1 = Time(h1, m1, s1)
        t2 = Time(h2, m2, s2)
        
        s1_total = t1.total_seconds()
        s2_total = t2.total_seconds()
        
        if s1_total < s2_total:
            print("Student 1")
        elif s2_total < s1_total:
            print("Student 2")
        else:
            print("Equal")
    except (ValueError, IndexError):
        pass

main()
```

Máy tính đã “chọn sẵn” một số bí mật S trong đoạn 1..100.
Bạn nhận được một danh sách các chuỗi mà người chơi đã nhập (có thể nhập sai, không phải số).
Nhiệm vụ: mô phỏng trò chơi đoán số theo quy tắc sau (theo đúng thứ tự các dòng nhập):
Với mỗi chuỗi x:
Nếu x không chuyển được sang số nguyên (gây ValueError) hoặc số nằm ngoài [1, 100] → in Invalid
Nếu chuyển được và value < S → in Too low
Nếu chuyển được và value > S → in Too high
Nếu value == S → in Correct và kết thúc ngay (không xử lý các dòng sau)
Nếu hết danh sách mà chưa đoán đúng, in thêm một dòng Game Over.
Yêu cầu bắt buộc: khi chuyển chuỗi sang số nguyên phải dùng try-except để bắt ValueError.
Input
Dòng 1: số nguyên S (1 ≤ S ≤ 100)
Dòng 2: số nguyên n — số lượt nhập
n dòng tiếp theo: mỗi dòng là một chuỗi người chơi nhập (có thể có khoảng trắng ở đầu/cuối)
Output
Với mỗi lượt (cho đến khi đúng hoặc hết lượt), in ra đúng một trong các dòng:
Invalid, Too low, Too high, Correct
Nếu không có Correct trong toàn bộ lượt, in thêm:
Game Over

```python
def guess_game():
    try:
        S = int(input())
        n = int(input())
        
        guessed_correctly = False
        for _ in range(n):
            line = input().strip()
            try:
                val = int(line)
                if val < 1 or val > 100:
                    print("Invalid")
                elif val < S:
                    print("Too low")
                elif val > S:
                    print("Too high")
                else:
                    print("Correct")
                    guessed_correctly = True
                    break
            except ValueError:
                print("Invalid")
                
        if not guessed_correctly:
            print("Game Over")
    except Exception:
        pass

guess_game()
```

Viết chương trình nhập vào một ký tự duy nhất từ bàn phím. Hãy xác định ký tự đó thuộc nhóm nào trong 4 nhóm sau:
- Chữ cái thường (a đến z)
- Chữ cái in hoa (A đến Z)
- Chữ số (0 đến 9)
- Các ký tự khác (ký tự đặc biệt, khoảng trắng, dấu câu...)
Dữ liệu vào (Input):
Gồm 1 dòng chứa đúng một ký tự c.
Dữ liệu ra (Output):
In ra LOWER nếu là chữ thường.
In ra UPPER nếu là chữ hoa.
In ra DIGIT nếu là chữ số.
In ra OTHER nếu là các loại ký tự khác.

```python
def classify_char():
    try:
        s = input()
        if not s:
            return
        c = s[0]
        
        if 'a' <= c <= 'z':
            print("LOWER")
        elif 'A' <= c <= 'Z':
            print("UPPER")
        elif '0' <= c <= '9':
            print("DIGIT")
        else:
            print("OTHER")
    except Exception:
        pass

classify_char()
```

Viết chương trình nhập vào một dãy số nguyên từ bàn phím (nhập từng số trên từng dòng). Quá trình nhập sẽ kết thúc ngay lập tức khi người dùng nhập vào số -1.
Hãy tính giá trị: K = Tổng các số chẵn - Tổng các số lẻ.
Lưu ý:
Số -1 là dấu hiệu kết thúc, không được tính vào tổng lẻ hay kết quả cuối cùng.
Số 0 được tính là số chẵn.
Dãy số có thể bao gồm số âm.
Dữ liệu vào (Input):
Các số nguyên x, mỗi số trên một dòng (hoặc nhập liên tiếp). Kết thúc bởi số -1.
Dữ liệu ra (Output):
Một số nguyên duy nhất là kết quả của phép trừ.
Một số nguyên duy nhất là kết quả của phép trừ.

```python
def main():
    even_sum = 0
    odd_sum = 0
    try:
        while True:
            line = input().strip()
            tokens = line.split()
            finish = False
            for token in tokens:
                val = int(token)
                if val == -1:
                    finish = True
                    break
                if val % 2 == 0:
                    even_sum += val
                else:
                    odd_sum += val
            if finish:
                break
    except (EOFError, ValueError):
        pass
    print(even_sum - odd_sum)

main()
```
